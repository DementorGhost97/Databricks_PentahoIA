### Lectura de archivos Pentaho Logs

```python
from pyspark.sql.functions import col

VOLUME_PATH ="/Volumes/pentaho_logs/bronze/volume_pentahologs"

archivos_path = [
    f"{VOLUME_PATH}/{archivo}"
    for archivo in archivos_nuevos
]

print("==== PATH PROXIMOS A PROCESARSE ====")

for path in archivos_path:
    print(path)



df_bronze_pentaho_logs = (
    spark.read
       .text(archivos_path)
       .select(
           col("value").alias("descripcion"),
        col("_metadata.file_path").alias("file_path"),
        col("_metadata.file_name").alias("file_name")
       )
    )
#display(df_bronze_pentaho_logs.limit(10))
```

### Enriquecimiento DataFrame pentaho log

```python

from pyspark.sql.functions import regexp_extract, to_date, current_timestamp,col,when
 
df_bronze_pentaho_logs = (df_bronze_pentaho_logs.withColumn("application",regexp_extract("file_name", r"^([A-Za-z]+).", 1))
                                        .withColumn("hora",regexp_extract("descripcion", r"(\d{2}:\d{2}:\d{2}\.\d{3})", 1))
                                        .withColumn("fecha",regexp_extract("descripcion",r"^(\d{4}-\d{2}-\d{2})",1))
                                        .withColumn("Nivel",regexp_extract("descripcion",r"\b(INFO|WARNING|SEVERE|DEBUG|ERROR)\b", 1))
                                        .withColumn("Proceso",regexp_extract("descripcion",r"-\s+([A-Za-z0-9_]+)(?:\.\d+)?\s+-",1))

                                        .withColumn("is_event_start",when(regexp_extract(col("descripcion"),  r"^\d{4}-\d{2}-\d{2}",0) != "",True).otherwise(False))
                                         .withColumn("Timestamp",current_timestamp())    

) 

                            

### limpiar data frame 
###df = df.drop("<campo>")
##df_bronze_catalina= df_bronze_catalina.drop("value")
#display(df_bronze_pentaho_logs)
```

#### Validaciones DATAFRAME

```python
# Número de registros
print(f"Total de líneas: {df_bronze_pentaho_logs.count():,}")

# Estructura
df_bronze_pentaho_logs.printSchema()
```

```python
from pyspark.sql.functions import col, sum, when

df_bronze_pentaho_logs.select(
    sum(when(col("descripcion").isNull(), 1).otherwise(0)).alias("descripcion_null"),
    sum(when(col("file_path").isNull(), 1).otherwise(0)).alias("file_path_null"),
    sum(when(col("file_name").isNull(), 1).otherwise(0)).alias("file_name_null"),
    sum(when(col("application").isNull(), 1).otherwise(0)).alias("application_null"),
 sum(when(col("hora").isNull(), 1).otherwise(0)).alias("hora"),
 sum(when(col("fecha").isNull(), 1).otherwise(0)).alias("fecha_null"),
  sum(when(col("Nivel").isNull(), 1).otherwise(0)).alias("Nivel_null"),
  sum(when(col("Proceso").isNull(), 1).otherwise(0)).alias("Proceso_null"),
).show()
```

### Creación tabla BRONZE_TABLE_LOGS_PENTAHO

```python
BRONZE_TABLE_PENTAHO_LOGS = "pentaho_logs.bronze.bronze_logs_pentaho"
(
    df_bronze_pentaho_logs.write
        .format("delta")
        .mode("append")
        .saveAsTable(BRONZE_TABLE_PENTAHO_LOGS)
)
```
